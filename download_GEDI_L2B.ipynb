```python
import earthaccess
earthaccess.login()

RoI = (104.64314981548942, 21.640988533288887, 104.82675553350205, 21.860378136937058)

granules = earthaccess.search_data(
    short_name="GEDI02_A",
    temporal=("2021-01-01", "2022-01-31"),
    bounding_box=RoI
    )

filename = earthaccess.download(granules[0], "../gedi_L2A")
```
